### Ejercicio 2: Datos climáticos multiestación

Crea un nuevo DataFrame con datos de 5 estaciones SENAMHI de diferentes regiones del Perú (https://www.senamhi.gob.pe/site/descarga-datos/).

Luego:
1. Paresear el nombres de la estaciones y leer los archivo.
1. Unir las db (Fecha, Est1, Est2, ...)
1. Calcula el promedio mensual de precipitación por región
2. Identifica el mes más lluvioso en cada estación
4. Exporta el resultado formateado a CSV

```python
import os
import pandas as pd

#1. Paresear el nombre de las estaciones y leer los archivos

# Directorio de los archivos
url = '/content/drive/MyDrive/9no ciclo/Senamhi-20260826T023849Z-1-001/Senamhi'

# Listar todos los archivos de la carpeta
files = os.listdir(url)

# Filtrar los 5 archivos
archivos_txt = [f for f in files if f.endswith('.txt')]

estaciones = {}

for station in archivos_txt:
    partes = station.replace('.txt', '').split('_', 1)
    codigo = partes[0]
    nombre = partes[1]

    # Leer el archivo de la estación
    station_df = pd.read_table(f'{url}/{station}', header=None, sep=' ', na_values=['-999', '-99.9', '-3256'] )

    # Asignar nombres de columnas
    station_df.columns = ['Año', 'Mes', 'Dia', 'Pr', 'Tmax', 'Tmin']
    station_df['Año'] = station_df['Año'].astype(int)
    station_df['Mes'] = station_df['Mes'].astype(int)
    station_df['Dia'] = station_df['Dia'].astype(int)
    station_df['Pr'] = station_df['Pr'].astype(float)
    station_df['Tmax'] = station_df['Tmax'].astype(float)
    station_df['Tmin'] = station_df['Tmin'].astype(float)
    # Guardar el DataFrame
    estaciones[nombre] = station_df

    print(f"Código: {codigo}, Nombre: {nombre}")
```

```python
#2. Unir las db (Fecha, Est1, Est2, ...)

lista_dfs = []

# Preparar cada DataFrame con el nombre de su estación en la columna de precipitación
for nombre, df in estaciones.items():
    df_sub = df[['Año', 'Mes', 'Dia', 'Pr']].rename(columns={'Pr': nombre})
    lista_dfs.append(df_sub)

# Unir progresivamente los 5 DataFrames por Año, Mes y Dia
clima_unificado = lista_dfs[0]
for df in lista_dfs[1:]:
    clima_unificado = pd.merge(clima_unificado, df, on=['Año', 'Mes', 'Dia'], how='outer')

# Crear la columna unificada 'Fecha' (Formato: Año-Mes-Día)
clima_unificado['Fecha'] = (clima_unificado['Año'].astype(str) + '-' + clima_unificado['Mes'].astype(str) + '-' +
clima_unificado['Dia'].astype(str))

# Reordenar las columnas para dejar 'Fecha' al inicio
estaciones_cols = list(estaciones.keys())
clima_unificado = clima_unificado[['Fecha', 'Año', 'Mes', 'Dia'] + estaciones_cols]

# Ordenar cronológicamente
clima_unificado = clima_unificado.sort_values(['Año', 'Mes', 'Dia']).reset_index(drop=True)

# Vista previa del resultado unificado
print(clima_unificado.head())
```

```python
#3. Calcula el promedio mensual de precipitación por región

estaciones_cols = list(estaciones.keys())

promedio_mensual = clima_unificado.groupby('Mes')[estaciones_cols].mean().round(4)

# Tabla de promedios mensuales
print("                 Promedio mensual de precipitación (mm):")
print(promedio_mensual)
```

```python
#4. Identifica el mes más lluvioso en cada estación
#Nombre de todos los meses
meses_nombres = {
    1: 'Enero', 2: 'Febrero', 3: 'Marzo', 4: 'Abril',
    5: 'Mayo', 6: 'Junio', 7: 'Julio', 8: 'Agosto',
    9: 'Septiembre', 10: 'Octubre', 11: 'Noviembre', 12: 'Diciembre'
}

mes_mas_lluvioso = promedio_mensual.idxmax().map(meses_nombres)

print("Mes más lluvioso por estación:")
print(mes_mas_lluvioso)
```

```python
#5. Exporta el resultado formateado a CSV

promedio_mensual.to_csv('promedio_mensual_estaciones.csv')
mes_mas_lluvioso.to_csv('mes_mas_lluvioso_estaciones.csv', header=['Mes_Mas_Lluvioso'])
print("\n¡Archivos CSV guardados correctamente!")
```
